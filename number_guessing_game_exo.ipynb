```python
import random
def number_guessing_game():
    random_number=random.randint(1,100)
    max_attemps=7
    print(random_number)
    print("welcome to the number guessing game")
    print("Im thinking of a number between 1 and 100")

    for i in range(1,max_attemps+1):
        guess=int(input("choisissez votre nombre"))

        if  guess<random_number :
            print ("Too low!")

        elif guess>random_number:
             print ("Too high!")
        else :
            print("congratulations you guessed the correct number")
            break
    else:
        print(f"Sorry, you've used all{max_attemps} attemps ")

number_guessing_game()
```

```python
user_word = input("Enter a word")
corrected_word = user_word[0]
for index in range(1,len(user_word)):
    if user_word[index] != user_word[index-1]:
        corrected_word += user_word[index]
print(f"user's word : {user_word} ➞ {corrected_word}")
```
