```python
# 📦 Étape 1 : Installation des bibliothèques nécessaires, fait ci-dessus
# !pip install -q langchain torch transformers sentence-transformers datasets faiss-cpu
# !pip install -U langchain-community

# 🧩 Étape 2 : Chargement du dataset Hugging Face
from langchain.document_loaders import HuggingFaceDatasetLoader

# Nom du dataset et colonne contenant le texte
dataset_name = "databricks/databricks-dolly-15k"
page_content_column = "context"

# Chargement des documents
loader = HuggingFaceDatasetLoader(dataset_name, page_content_column)
data = loader.load()
print(data[:2])  # Affiche les deux premiers documents pour vérification

# ✂️ Étape 3 : Découpage des documents en chunks
from langchain.text_splitter import RecursiveCharacterTextSplitter

# Création du splitter avec chevauchement
text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=150)
docs = text_splitter.split_documents(data)
print(docs[0])  # Affiche le premier chunk

# 🔢 Étape 4 : Création des embeddings avec un modèle sentence-transformers
from langchain.embeddings import HuggingFaceEmbeddings

# Configuration du modèle
modelPath = "sentence-transformers/all-MiniLM-l6-v2"
model_kwargs = {'device': 'cpu'}
encode_kwargs = {'normalize_embeddings': False}

# Initialisation des embeddings
embeddings = HuggingFaceEmbeddings(
    model_name=modelPath,
    model_kwargs=model_kwargs,
    encode_kwargs=encode_kwargs
)

# Test d'embedding
text = "This is a test document."
query_result = embeddings.embed_query(text)
print(query_result[:3])  # Affiche les 3 premières valeurs de l'embedding

# 🗃️ Étape 5 : Création du vecteur store FAISS
from langchain.vectorstores import FAISS

# Indexation des documents
db = FAISS.from_documents(docs, embeddings)

# 🤖 Étape 6 : Préparation du modèle de langage (LLM)
from transformers import AutoTokenizer, AutoModelForQuestionAnswering, pipeline
from langchain import HuggingFacePipeline

# Chargement du tokenizer et du modèle
model_name = "Intel/dynamic_tinybert"
tokenizer = AutoTokenizer.from_pretrained(model_name, padding=True, truncation=True, max_length=512)
model = AutoModelForQuestionAnswering.from_pretrained(model_name)

# Création du pipeline de question-réponse
Youtubeer = pipeline(
    "question-answering",
    model=model,
    tokenizer=tokenizer,
    return_tensors='pt'
)

# Intégration dans Langchain
llm = HuggingFacePipeline(
    pipeline=Youtubeer,
    model_kwargs={"temperature": 0.7, "max_length": 512},
)
```

```python
# 🔗 Étape 7 : Construction de la chaîne RetrievalQA
from langchain.chains import RetrievalQA

# Création du retriever
retriever = db.as_retriever(search_kwargs={"k": 4})

# Construction de la chaîne QA
qa = RetrievalQA.from_chain_type(
    llm=llm,
    chain_type="refine",
    retriever=retriever,
    return_source_documents=False
)

# ❓ Étape 8 : Test du système RAG
question = "What is cheesemaking?"
result = qa.run({"query": question})
print(result)  # Affiche la réponse générée
```

```python
# 🤖 Étape 6 corrigée : Utiliser un modèle de génération
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM, pipeline
from langchain import HuggingFacePipeline

# Choix d’un modèle de génération compatible
model_name = "google/flan-t5-base"

# Chargement du tokenizer et du modèle
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

# Création du pipeline de génération
generator = pipeline(
    "text2text-generation",
    model=model,
    tokenizer=tokenizer,
    max_length=512
)

# Intégration dans Langchain
llm = HuggingFacePipeline(
    pipeline=generator,
    model_kwargs={"temperature": 0.7, "max_length": 512},
)

# 🔗 Étape 7 : Construction de la chaîne RetrievalQA
from langchain.chains import RetrievalQA

retriever = db.as_retriever(search_kwargs={"k": 4})

qa = RetrievalQA.from_chain_type(
    llm=llm,
    chain_type="refine",
    retriever=retriever,
    return_source_documents=False
)

# ❓ Étape 8 : Test du système RAG
question = "What is cheesemaking?"
result = qa.run(question)
print(result)
```
