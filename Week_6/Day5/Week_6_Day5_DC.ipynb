```python
# 📦 Importation des bibliothèques nécessaires
import pandas as pd
import numpy as np
import re
import networkx as nx
import spacy
from nltk.corpus import stopwords
from sklearn.metrics.pairwise import cosine_similarity


# 📁 Chargement du fichier CSV
df = pd.read_csv(
    '/content/drive/MyDrive/Colab Notebooks/W6/Day_5/tennis_articles.csv',
    encoding='latin1'
)

df.drop(columns=['article_title'], inplace=True)  # 🧹 Suppression du titre
# ✂️ Tokenisation des phrases avec spaCy
sentences = []
for text in df['article_text']:
    doc = nlp(text)
    sentences += [sent.text.strip() for sent in doc.sents]  # on extrait les phrases nettoyées

print(f"Nombre total de phrases : {len(sentences)}")

# 📚 Chargement des vecteurs GloVe (100 dimensions)
glove_file = '/content/glove.6B.100d.txt'

embeddings = {}
with open(glove_file, 'r', encoding='utf-8') as f:
    for line in f:
        values = line.split()
        word = values[0]
        vector = np.array(values[1:], dtype='float32')
        embeddings[word] = vector

print(f"Nombre de mots dans GloVe : {len(embeddings)}")

# 🧹 Nettoyage et normalisation du texte
stop_words = set(stopwords.words('english'))

def nettoyer_phrase(phrase):
    phrase = re.sub(r'[^\w\s]', '', phrase)  # retire ponctuation
    phrase = re.sub(r'\d+', '', phrase)      # retire les chiffres
    phrase = phrase.lower()
    mots = phrase.split()
    mots = [mot for mot in mots if mot not in stop_words]
    return mots

# 🔤 Vectorisation des phrases
def vectoriser(mots):
    vecteurs = [embeddings.get(mot, np.zeros((100,))) for mot in mots]
    if vecteurs:
        return np.mean(vecteurs, axis=0)
    else:
        return np.zeros((100,))

sentence_vectors = [vectoriser(nettoyer_phrase(sentence)) for sentence in sentences]

# 🧠 Construction de la matrice de similarité
n = len(sentences)
similarity_matrix = np.zeros((n, n))

for i in range(n):
    for j in range(n):
        if i != j:
            similarity_matrix[i][j] = cosine_similarity(
                sentence_vectors[i].reshape(1, 100),
                sentence_vectors[j].reshape(1, 100)
            )[0][0]

# 🔗 Création du graphe et application du PageRank
nx_graph = nx.from_numpy_array(similarity_matrix)
scores = nx.pagerank(nx_graph)

# 📌 Sélection des phrases les mieux classées
top_n = 10
ranked_sentences = sorted(
    ((scores[i], s) for i, s in enumerate(sentences)),
    reverse=True
)

print("📝 Résumé automatique des articles :\n")
for i in range(top_n):
    print(f"- {ranked_sentences[i][1]}")

```
