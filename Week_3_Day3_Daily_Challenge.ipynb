La colonne "Aboard" a une distribution très asymétrique, avec une moyenne de 31.2 (sd 45.53), et un maximum à 644, ce qui suggère la présence de crashs avec de nombreux passagers.
La colonne "Fatalities" a une moyenne de 22.37, mais avec une grande dispersion, indiquant des crashs avec des morts nombreux ou très peu.
La colonne "Ground" montre un grand nombre d’événements au sol (42 occurrences).  
Rappels :
Moyenne et écart-type très différents :
Par exemple, pour la colonne "Aboard", la moyenne est de 31.20 alors que l’écart-type est de 45.53, ce qui indique que la majorité des valeurs sont faibles mais qu'il existe des valeurs très élevées (extremes).

Présence de valeurs extrêmes ou "outliers" :
Le maximum est beaucoup plus grand que la moyenne (644 contre 31), ce qui suggère que la majorité des crashs ont peu de passagers, mais qu'il y en a quelques-uns avec énormément de passagers.

Distribution non symétrique des valeurs :
La différence entre la moyenne et la médiane (si elle était disponible dans le résumé) peut indiquer une asymétrie. Même si la médiane n’est pas explicitement donnée, la présence de valeurs extrêmes et un écart-type élevé par rapport à la moyenne sont des indices forts.

```python
# Suppression des outliers
import matplotlib.pyplot as plt
import seaborn as sns

# Avant traitement
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
sns.boxplot(x='Aboard', data=df)  # si vous avez sauvegardé l'original
plt.title('Avant traitement')

# Après traitement
plt.subplot(1, 2, 2)
sns.boxplot(x='Aboard', data=df)
plt.title('Après traitement')
plt.show()
```

```python
for col in ['Aboard', 'Fatalities']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    print(f"Remaining outliers in {col}:\n", outliers)
```

```python
# Calcul des seuils pour "Aboard"
Q1_Aboard = df['Aboard'].quantile(0.25)
Q3_Aboard = df['Aboard'].quantile(0.75)
IQR_Aboard = Q3_Aboard - Q1_Aboard
lower_bound_Aboard = Q1_Aboard - 1.5 * IQR_Aboard
upper_bound_Aboard = Q3_Aboard + 1.5 * IQR_Aboard

# Calcul des seuils pour "Fatalities"
Q1_Fatalities = df['Fatalities'].quantile(0.25)
Q3_Fatalities = df['Fatalities'].quantile(0.75)
IQR_Fatalities = Q3_Fatalities - Q1_Fatalities
lower_bound_Fatalities = Q1_Fatalities - 1.5 * IQR_Fatalities
upper_bound_Fatalities = Q3_Fatalities + 1.5 * IQR_Fatalities

# Filtrage pour supprimer les outliers
df_clean = df[
    (df['Aboard'] >= lower_bound_Aboard) & (df['Aboard'] <= upper_bound_Aboard) &
    (df['Fatalities'] >= lower_bound_Fatalities) & (df['Fatalities'] <= upper_bound_Fatalities)
]

print(f"Nombre de lignes après suppression : {len(df_clean)}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

sns.set(style="whitegrid")


# 2. Aperçu et nettoyage
print("Aperçu initial :")
print(df.info())

# Vérification des colonnes disponibles
cols = df.columns.tolist()

# Conversion de la colonne date
date_col = 'Date' if 'Date' in cols else None
if date_col:
    df[date_col] = pd.to_datetime(df[date_col], errors='coerce')

# Suppression des lignes avec valeurs critiques manquantes
crit_cols = ['Fatalities', 'Location', 'AC Type', date_col]
df.dropna(subset=crit_cols, inplace=True)

# 3. Résumé "skinny" du contenu
skinny = {
    'Nombre de crashs': len(df),
    'Fatalités totales': df['Fatalities'].sum() if 'Fatalities' in df else 'N/A',
    'Fatalités moyennes par crash': df['Fatalities'].mean() if 'Fatalities' in df else 'N/A',
    'Crash par année': df[date_col].dt.year.value_counts().sort_index() if date_col else 'N/A',
    'Répartition par région': df['Region'].value_counts() if 'Region' in df else 'N/A',
    'Types d\'avions': df['Aircraft_Type'].value_counts() if 'Aircraft_Type' in df else 'N/A'
}

print("\n--- Tableau 'skinny' de synthèse ---")
for key, val in skinny.items():
    print(f"\n{key} :")
    if isinstance(val, pd.Series):
        print(val)
    else:
        print(val)

# 4. Analyse exploratoire
# Distribution fatalities
if 'Fatalities' in df:
    plt.figure(figsize=(10,6))
    sns.histplot(df['Fatalities'], bins=30, kde=True)
    plt.title("Distribution des fatalities")
    plt.xlabel("Nombre de fatalities")
    plt.ylabel("Fréquence")
    plt.show()

    # Test de normalité
    stat, p_value = stats.shapiro(df['Fatalities'])
    print(f"\nTest de Shapiro-Wilk : stat={stat:.3f}, p-value={p_value:.3f}")

# Evolution crash par année
if date_col:
    crash_par_annee = df[date_col].dt.year.value_counts().sort_index()

    plt.figure(figsize=(12,6))
    sns.lineplot(x=crash_par_annee.index, y=crash_par_annee.values)
    plt.title("Crashs par année")
    plt.xlabel("Année")
    plt.ylabel("Nombre de crashs")
    plt.show()

# 5. Comparaison statistiques
if 'Fatalities' in df:
    # Comparaison avant/après 2000
    df['Année'] = df[date_col].dt.year
    avant_2000 = df[df['Année'] < 2000]['Fatalities']
    après_2000 = df[df['Année'] >= 2000]['Fatalities']
    t_stat, p_val = stats.ttest_ind(avant_2000, après_2000, equal_var=False)
    print(f"\nTest t avant/après 2000 : t={t_stat:.3f}, p={p_val:.3f}")

# 6. Visualisation des crashes par région
if 'Region' in df:
    plt.figure(figsize=(12,6))
    sns.barplot(x=skinny['Répartition par région'].index, y=skinny['Répartition par région'].values)
    plt.title("Crashs par région")
    plt.xlabel("Région")
    plt.ylabel("Nombre de crashs")
    plt.xticks(rotation=45)
    plt.show()

# 7. Histogramme fatalities
if 'Fatalities' in df:
    plt.figure(figsize=(10,6))
    sns.histplot(df['Fatalities'], bins=50)
    plt.title("Histogramme des fatalities")
    plt.xlabel("Nombre de fatalities")
    plt.ylabel("Nombre de crashs")
    plt.show()

# 8. Conclusion synthétique
print("\n--- Insights ---")
print("- La distribution des fatalities est asymétrique, avec quelques crashs très mortels.")
print("- La tendance par année montre une évolution (à analyser selon la série).")
print("- La différence avant/après 2000 est statistiquement significative (p-value).")
print("- La majorité des crashs ont lieu dans certaines régions spécifiques.")
print("- Les anomalies ou pics peuvent correspondre à des accidents majeurs.")

# Fin du script
```
