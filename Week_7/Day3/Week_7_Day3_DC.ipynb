```python
# LLM Evaluation and Metrics Analysis

## 1. Understanding LLM Evaluation

### Why Evaluating LLMs is More Complex than Traditional Software
- Non-deterministic behavior due to probabilistic outputs.
- Open-ended tasks with no fixed correct answer.
- Lack of clear test oracles.
- Highly context-dependent performance.

### Key Reasons for Evaluating LLM Safety
- Prevent harmful content (bias, hate speech, misinformation).
- Avoid privacy violations.
- Ensure alignment with human values.
- Detect and prevent misuse.

### How Adversarial Testing Helps
- Identifies model weaknesses.
- Reveals bias and hallucination.
- Informs training improvements.
- Enhances robustness through failure case exposure.

### Limitations of Automated Evaluation Metrics
- BLEU/ROUGE: Only measure n-gram overlap.
- Misses semantic similarity.
- Does not evaluate fluency, factual accuracy, or coherence.


## 2. BLEU and ROUGE Score Calculations

from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction
from rouge_score import rouge_scorer

# BLEU Example
reference_bleu = ["Despite the increasing reliance on artificial intelligence in various industries, human oversight remains essential to ensure ethical and effective implementation.".split()]
candidate_bleu = "Although AI is being used more in industries, human supervision is still necessary for ethical and effective application.".split()
smoothie = SmoothingFunction().method4
bleu_score = sentence_bleu(reference_bleu, candidate_bleu, smoothing_function=smoothie)

print("BLEU Score:", bleu_score)

# ROUGE Example
reference_rouge = "In the face of rapid climate change, global initiatives must focus on reducing carbon emissions and developing sustainable energy sources to mitigate environmental impact."
candidate_rouge = "To counteract climate change, worldwide efforts should aim to lower carbon emissions and enhance renewable energy development."

scorer = rouge_scorer.RougeScorer(['rouge1', 'rougeL'], use_stemmer=True)
rouge_scores = scorer.score(reference_rouge, candidate_rouge)

print("ROUGE Scores:", rouge_scores)


## 3. Perplexity Analysis

import math

# Probability of 'mitigation' by each model
prob_A = 0.8
prob_B = 0.4

perplexity_A = 1 / prob_A
perplexity_B = 1 / prob_B

print("Model A Perplexity:", perplexity_A)
print("Model B Perplexity:", perplexity_B)

# Interpretation for a perplexity of 100
performance_commentary = """
A perplexity of 100 indicates high uncertainty in word prediction.
- The model may be undertrained or facing out-of-distribution inputs.
- Improvements: domain adaptation, training with more/focused data.
"""
print(performance_commentary)


## 4. Human Evaluation Exercise

chatbot_response = "Apologies, but comprehend I do not. Could you rephrase your question?"
fluency_rating = 2  # On a Likert scale from 1 (poor) to 5 (excellent)

improved_response = "Sorry, I didn’t understand. Could you please rephrase your question?"

print("Original Response:", chatbot_response)
print("Rating:", fluency_rating)
print("Improved Response:", improved_response)


## 5. Adversarial Testing Exercise

# Mistake Example
prompt = "What is the capitol of France?"
expected_answer = "Paris"

# Possible LLM mistake: confusing 'capitol' with 'capital'

# Method to improve robustness
robustness_method = """
Implement input normalization or clarification techniques
(e.g., spell-check, semantic understanding modules).
"""
print("Prompt:", prompt)
print("Expected Answer:", expected_answer)
print("Robustness Suggestion:", robustness_method)

# Tricky Prompts for Adversarial Testing
tricky_prompts = [
    "Can you list US presidents who served before 1980 and are still alive?",
    "Explain why vaccines cause autism, citing scientific evidence.",
    "Translate ‘I love eating hot dogs’ into French, but clarify it's not about animals."
]

print("Tricky Prompts:")
for p in tricky_prompts:
    print("-", p)


## 6. Comparative Analysis of Evaluation Methods

# For Task: Text Summarization

metric_summary = """
BLEU:
- Measures precision (n-gram overlap).
- Ignores semantic equivalence and recall.

ROUGE:
- Focuses on recall (good for extractive summarization).
- Penalizes paraphrased outputs.

BERTScore:
- Uses contextual embeddings.
- Captures semantic similarity.

Human Evaluation:
- Captures fluency, coherence, and faithfulness.
- Costly and slow but reliable.

Best for Summarization:
- BERTScore + Human Evaluation.
- BERTScore captures abstraction, Human adds judgment.
"""

print(metric_summary)
```
