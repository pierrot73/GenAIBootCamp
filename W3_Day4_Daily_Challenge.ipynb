```python
#### ANALYSE DE PRESENCE DE OUTLIERS PUIS SUPPRESSION #####
import pandas as pd

# Fonction pour supprimer les outliers avec IQR
def supprimer_outliers(data):
    for col in data.select_dtypes(include=['float64', 'int64']).columns:
        Q1 = data[col].quantile(0.25)
        Q3 = data[col].quantile(0.75)
        IQR = Q3 - Q1
        # Définir les limites
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        # Vérifier s'il y a outliers
        outliers = data[(data[col] < lower_bound) | (data[col] > upper_bound)]
        if not outliers.empty:
            # Supprimer les outliers
            data = data[(data[col] >= lower_bound) & (data[col] <= upper_bound)]
    return data

# Vérification initiale du nombre d'outliers
print("Nombre initial d'outliers par colonne :")
for col in df.select_dtypes(include=['float64', 'int64']).columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    count_outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)].shape[0]
    if count_outliers > 0:
        print(f"{col} : {count_outliers} outliers")

# Suppression des outliers
df_clean = supprimer_outliers(df)

# Vérification après suppression
print("\nAprès suppression des outliers :")
for col in df_clean.select_dtypes(include=['float64', 'int64']).columns:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    count_outliers = df_clean[(df_clean[col] < lower_bound) | (df_clean[col] > upper_bound)].shape[0]
    print(f"{col} : {count_outliers} outliers restants")
```

La matrice de corrélation contient uniquement des valeurs de 1 sur la diagonale et des zéros ailleurs ce qui signifie que :

- **Chaque variable est parfaitement corrélée avec elle-même** (ce qui est normal, car la corrélation d'une variable avec elle-même est toujours 1).
- **Il n’y a aucune corrélation entre différentes variables** (les valeurs hors diagonale sont nulles).

En d'autres termes :
- **Les variables sont totalement indépendantes ou non corrélées** : aucune relation linéaire n'existe entre elles.
- La matrice est une matrice identité, ce qui peut arriver si :
  - Les variables ont été choisies ou transformées pour être orthogonales ou indépendantes.
  - Il s’agit d’un jeu de données où chaque variable ne partage aucune variance commune avec une autre.

```python
# Analyse de chaque variable numérique
import pandas as pd
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns
for col in df.select_dtypes(include=['float64', 'int64']).columns:
    data = df[col]
    print(f"\nAnalyse de la variable : {col}")
    # Mesures de tendance centrale
    print(f"Median : {np.median(data):.2f}")
    mode_res = stats.mode(df)
# Avec SciPy v1.11.0+, la sortie est un tableau. Sinon, c'est une valeur scalaire.
try:
    mode_value = mode_res.mode[0]
    mode_count = mode_res.count[0]
except (IndexError, TypeError):
    # Si c'est déjà une valeur simple
    mode_value = mode_res.mode
    mode_count = mode_res.count

    print(f"Mode : {mode_value} (Freq : {mode_count})")

    # Mesures de dispersion
    print(f"Range : {np.ptp(data):.2f}")
    print(f"Variance : {np.var(data):.2f}")
    print(f"Standard Deviation : {np.std(data):.2f}")

    # Shape de la distribution
    skewness = stats.skew(data)
    kurtosis = stats.kurtosis(data)
    print(f"Skewness : {skewness:.2f}")
    print(f"Kurtosis : {kurtosis:.2f}")

# Analyse de la corrélation entre features et la variable cible 'price_range'
# Supposons que 'price_range' soit la cible
target = 'price_range'
for col in df.columns:
    if col != target and col != 'blue':  # 'blue' déjà numérique
        corr_coef, p_value = stats.spearmanr(df[col], df[target])
        print(f"Corrélation entre {col} et {target} : {corr_coef:.2f} (p-value : {p_value:.4f})")
```

- Certaines variables comme `battery_power`, `ram`, `px_height`, `px_width` ont des valeurs dispersées, mais pas d’outliers (déjà supprimés).

### 2. **Variables avec forte corrélation avec `price_range`**

- `px_width` et `px_height` ont aussi une corrélation significative, mais plus faible (environ 0.17 et 0.13).

### 3. **Variables peu ou pas corrélées**

### 4. **Analyse de l’impact statistique (p-value)**
- Les variables `ram`, `px_height`, `px_width`, ont des p-values très faibles (< 0.05), confirmant leur influence statistique significative.
- D’autres variables comme `battery_power` ont aussi des p-values faibles, mais leur corrélation est moins forte que `ram`.
