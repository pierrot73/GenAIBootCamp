```python
# Importation des bibliothèques nécessaires
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from datetime import datetime

# Conversion de la colonne 'Date' en datetime
df['Date'] = pd.to_datetime(df['Date'], format='%d/%m/%Y')

# Vérification de la fréquence des données
date_diffs = df['Date'].diff().dropna()
print("\nDifferences entre les dates (premiers 5) :")
print(date_diffs.head())

# Fréquence la plus courante (en jours)
freq_days = date_diffs.mode()[0]
print(f"\nFréquence la plus courante (en jours) entre les observations : {freq_days}")

# 2. Visualisation des données
# Tracer le prix de clôture et le volume au fil du temps
plt.figure(figsize=(14,6))
plt.subplot(2,1,1)
plt.plot(df['Date'], df['Close'], label='Prix de clôture', color='blue')
plt.title("Prix de clôture au fil du temps")
plt.xlabel("Date")
plt.ylabel("Prix ($)")
plt.legend()

plt.subplot(2,1,2)
plt.plot(df['Date'], df['Volume'], label='Volume', color='orange')
plt.title("Volume échangé au fil du temps")
plt.xlabel("Date")
plt.ylabel("Volume")
plt.legend()

plt.tight_layout()
plt.show()

# Création manuelle d'un graphique en chandelier (candlestick) sans mplfinance
# Préparer les données
df_ohlc = df[['Date', 'Open', 'High', 'Low', 'Close']].copy()
df_ohlc['Date_num'] = df_ohlc['Date'].map(mdates.date2num)

# Définir la taille de chaque bougie (en jours)
width = 0.6  # largeur de la bougie en unité de date
color_up = 'g'
color_down = 'r'

fig, ax = plt.subplots(figsize=(15,7))

for index, row in df_ohlc.iterrows():
    date = row['Date_num']
    open_price = row['Open']
    high_price = row['High']
    low_price = row['Low']
    close_price = row['Close']

    # Déterminer la couleur de la bougie
    if close_price >= open_price:
        color = color_up
        lower = open_price
        height = close_price - open_price
    else:
        color = color_down
        lower = close_price
        height = open_price - close_price

    # Tracer la "tige" (wick)
    ax.plot([date, date], [low_price, high_price], color='black', linewidth=1)

    # Tracer le corps de la bougie
    rect = plt.Rectangle(
        (date - width/2, lower),  # position x, y
        width,                   # largeur
        height,                  # hauteur
        facecolor=color,
        edgecolor='black'
    )
    ax.add_patch(rect)

# Formatage de l'axe des dates
ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m'))
ax.xaxis.set_major_locator(mdates.MonthLocator(interval=3))
ax.set_title("Graphique en chandeliers (candlestick) sans mplfinance")
ax.set_xlabel('Date')
ax.set_ylabel('Prix ($)')

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# 3. Visualisation des distributions (histogrammes) pour chaque variable numérique
for col in ['Open', 'High', 'Low', 'Close', 'Adj Close', 'Volume']:
    plt.figure(figsize=(6, 4))
    sns.histplot(df[col], bins=50, kde=True)
    plt.title(f'Distribution de {col}')
    plt.xlabel(col)
    plt.ylabel('Fréquence')
    plt.show()

# 4. Matrice de corrélation
corr_matrix = df[['Open', 'High', 'Low', 'Close', 'Adj Close', 'Volume']].corr()
plt.figure(figsize=(8,6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Matrice de corrélation entre variables financières")
plt.show()

# 5. Tests de normalité pour chaque variable numérique
print("\nTests de normalité (Shapiro-Wilk):")
for col in ['Open', 'High', 'Low', 'Close', 'Adj Close', 'Volume']:
    stat, p_value = stats.shapiro(df[col])
    print(f"{col} : stat={stat:.3f}, p-value={p_value:.3f}")
    if p_value > 0.05:
        print(f"  -> {col} suit probablement une distribution normale.")
    else:
        print(f"  -> {col} ne suit probablement pas une distribution normale.")

# 6. Comparaison des moyennes entre deux périodes (exemple)
# Définir deux périodes, par exemple avant et après une date clé
date_split = '1985-01-01'  # Modifier selon votre contexte
df['Date'] = pd.to_datetime(df['Date'])
df_period1 = df[df['Date'] < date_split]
df_period2 = df[df['Date'] >= date_split]

# Test t pour la différence de moyenne de 'Close'
t_stat, p_val = stats.ttest_ind(df_period1['Close'], df_period2['Close'], equal_var=False)
print(f"\nTest t pour 'Close' entre deux périodes : t={t_stat:.3f}, p-value={p_val:.3f}")
if p_val < 0.05:
    print("La différence de moyenne est statistiquement significative.")
else:
    print("Aucune différence statistiquement significative.")
```

Les résultats montrent que toutes les variables numériques, y compris `Open`, `High`, `Low`, `Close`, `Adj Close` et `Volume`, ne suivent pas une distribution normale (p-value très faible). La différence significative de `Close` entre deux périodes indique aussi un changement notable.

```python
#### ANALYSE DE PRESENCE DE OUTLIERS PUIS SUPPRESSION #####
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Liste des colonnes à analyser
colonnes = ['Open', 'High', 'Low', 'Close', 'Adj Close', 'Volume']

# Fonction pour détecter et afficher les outliers avec boxplot
def detecter_outliers(df, colonne):
    plt.figure(figsize=(6, 4))
    sns.boxplot(x=df[colonne])
    plt.title(f'Boxplot pour {colonne}')
    plt.show()

    # Calcul des quartiles
    Q1 = df[colonne].quantile(0.25)
    Q3 = df[colonne].quantile(0.75)
    IQR = Q3 - Q1

    # Définir les limites pour détecter les outliers
    borne_inferieure = Q1 - 1.5 * IQR
    borne_superieure = Q3 + 1.5 * IQR

    # Identifier les outliers
    outliers = df[(df[colonne] < borne_inferieure) | (df[colonne] > borne_superieure)]
    print(f"{len(outliers)} outliers détectés dans {colonne}")

    return outliers

# Dictionnaire pour stocker tous les outliers détectés
outliers_total = pd.DataFrame()

# Boucle sur chaque colonne pour détecter les outliers
for col in colonnes:
    outliers_col = detecter_outliers(df, col)
    outliers_total = pd.concat([outliers_total, outliers_col])

# Si des outliers ont été détectés, les supprimer
if not outliers_total.empty:
    print(f"\nSuppression de {len(outliers_total)} outliers...")
    # On supprime toutes les lignes où au moins une colonne est un outlier
    df_sans_outliers = df.drop(outliers_total.index).reset_index(drop=True)
else:
    print("Aucun outlier détecté, pas de suppression.")

# Vérification finale
print("\nRésumé après suppression des outliers :")
print(df_sans_outliers.describe())

# Optionnel : visualiser à nouveau pour confirmer
for col in colonnes:
    plt.figure(figsize=(6, 4))
    sns.boxplot(x=df_sans_outliers[col])
    plt.title(f'Boxplot après suppression pour {col}')
    plt.show()
```
