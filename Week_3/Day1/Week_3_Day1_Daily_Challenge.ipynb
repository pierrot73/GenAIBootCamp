```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Fixer la reproductibilité des nombres aléatoires
np.random.seed(0)

# Définir les noms des villes du dataset
cities = ['Asadabad', 'Fayzabad', 'Jalalabad', 'Kunduz', 'Qala i Naw',
          'Sheberghan', 'Peshkopi', 'Shkodër', 'Tirana', 'Adrar']

# Définir les mois
months = ['Janvier', 'Février', 'Mars', 'Avril', 'Mai', 'Juin',
          'Juillet', 'Août', 'Septembre', 'Octobre', 'Novembre', 'Décembre']

# 1. Préparation des données
# Générer une matrice de températures aléatoires entre -5 et 35 degrés Celsius
# pour 10 villes et 12 mois
# np.random.uniform(low, high, size) pour générer les données
temperature_data = np.random.uniform(low=-5, high=35, size=(10, 12))

# Créer un DataFrame à partir de ces données
# Les villes seront l'index, les mois seront les colonnes
temperature_df = pd.DataFrame(data=temperature_data, index=cities, columns=months)

# Afficher les premières lignes pour vérification
print("Données de température (extrait):")
print(temperature_df.head(10))

# 2. Analyse des données
# Calculer la température moyenne annuelle pour chaque ville
annual_avg_temps = temperature_df.mean(axis=1)

# Identifier la ville avec la température moyenne la plus élevée
city_hottest = annual_avg_temps.idxmax()
temp_hottest = annual_avg_temps.max()

# Identifier la ville avec la température moyenne la plus basse
city_coldest = annual_avg_temps.idxmin()
temp_coldest = annual_avg_temps.min()

# Afficher les résultats
print(f"\nVille avec la température moyenne annuelle la plus élevée : {city_hottest} ({temp_hottest:.2f}°C)")
print(f"Ville avec la température moyenne annuelle la plus basse : {city_coldest} ({temp_coldest:.2f}°C)")

# 3. Visualisation
# Tracer les tendances mensuelles de température pour chaque ville
plt.figure(figsize=(12, 6))
for city in cities:
    plt.plot(months, temperature_df.loc[city], marker='o', label=city)

plt.title('Tendances mensuelles de température pour chaque ville')
plt.xlabel('Mois')
plt.ylabel('Température (°C)')
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```
